```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_athletes = pd.read_json('../dataset/olympic_athletes.json')
df_athletes.head()
```

```python
print(df_athletes.shape)
print(df_athletes.dtypes)
df_athletes.isnull().sum()
```

```python
percent_missing = df_athletes.isnull().sum() * 100 / len(df_athletes)
print(percent_missing.sort_values(ascending=False))
```

```python
# supprimer colonnes
df_athletes = df_athletes.drop(['athlete_medals', 'bio'], axis=1)
```

```python
# Remplire na
df_athletes['athlete_year_birth'] = df_athletes['athlete_year_birth'].fillna(df_athletes['athlete_year_birth'].median())

df_athletes['first_game'] = df_athletes['first_game'].fillna(df_athletes['first_game'].mode()[0])
```

```python
df_medals = pd.read_excel('../dataset/olympic_medals.xlsx')
df_medals.head()
```

```python
print(df_medals.shape)
print(df_medals.dtypes)
df_medals.isnull().sum()
```

```python
percent_missing = df_medals.isnull().sum() * 100 / len(df_medals)
print(percent_missing.sort_values(ascending=False))
```

```python
# Regarde les lignes GameTeam avec et sans URL
gameteam = df_medals[df_medals['participant_type'] == 'GameTeam']
print(gameteam['athlete_url'].isnull().sum())
print(gameteam['athlete_url'].notnull().sum())
```

```python
mask = df_medals['athlete_url'].isnull()
print(df_medals[mask]['participant_type'].value_counts())
```
